```python
import math

DEPTH = 1024
WIDTH = 12
AMP = (2**(WIDTH - 1)) - 1  # 2047 para 12 bits signed

def to_twos_complement(value: int, bits: int) -> int:
    if value < 0:
        value = (1 << bits) + value
    return value

with open("cos_lut.mif", "w") as f:
    f.write(f"WIDTH={WIDTH};\n")
    f.write(f"DEPTH={DEPTH};\n\n")
    f.write("ADDRESS_RADIX=UNS;\n")
    f.write("DATA_RADIX=HEX;\n\n")
    f.write("CONTENT BEGIN\n")

    for addr in range(DEPTH):
        angle = 2.0 * math.pi * addr / DEPTH
        value = round(AMP * math.cos(angle))   # faixa: -2047 a +2047
        value_tc = to_twos_complement(value, WIDTH)
        f.write(f"{addr:4d} : {value_tc:03X};\n")

    f.write("END;\n")

print("Arquivo cos_lut.mif gerado com sucesso.")
```
